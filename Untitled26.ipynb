```python
import pandas as pd
df = pd.read_csv('Cleaned_HR_Data_Analysis.csv')

df.head()
df.info()
df.describe()
```

```python
df["Age"] = df["Age"].astype(int)
df.head()
```

```python
def limpar_dados(df):
    df = df.copy()


    df = df.drop_duplicates()


    df = df.fillna(0)


    df["DateofHire"] = pd.to_datetime(df["DateofHire"])
    df["DateofTermination"] = pd.to_datetime(df["DateofTermination"], errors='coerce')

    return df
```

```python
def criar_features(df):
    df = df.copy()

    # tempo de empresa
    df["tempo_empresa"] = (
        pd.Timestamp("today") - df["DateofHire"]
    ).dt.days // 365

    # turnover
    df["attrition"] = df["DateofTermination"].notna().astype(int)

    # faixa salarial
    df["faixa_salarial"] = pd.qcut(df["Salary"], 4, labels=["baixo", "medio", "alto", "top"])

    return df
```

```python
def turnover(df):
    demissoes = df["attrition"].sum()
    total = len(df)
    return demissoes / total


def custo_total(df):
    return df["Salary"].sum()


def custo_medio(df):
    return df["Salary"].mean()
```
